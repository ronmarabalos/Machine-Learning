```
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Loads the newest export in data/
path = sorted(glob.glob('../data/*.csv'))[-1]
df = pd.read_csv(path)
print(path, df.shape)
df.head()
```

## Clean
The export has: `Date, Type, Amount, Category, Account, Description`

```
df['Date'] = pd.to_datetime(df['Date'])
df['Amount'] = pd.to_numeric(df['Amount'])
df['Description'] = df['Description'].fillna('').str.strip()
expenses = df[df['Type'] == 'expense']

df.info()
print(len(expenses), 'expenses')
```

## Q1. What are my top 5 spending categories?
Hint: `groupby('Category')`, `.sum()`, `.sort_values()`, then `.plot.barh()`

```
# TODO
top_5 = (
  df[df['Type'] == 'expense']
  .groupby('Category')['Amount']
  .sum()
  .sort_values(ascending=False)
  .head(5)
)
print(top_5)
```

## Q2. How much did I spend each month?
Hint: `expenses.groupby(expenses['Date'].dt.to_period('M'))['Amount'].sum()`, then `.plot()`

```
# TODO
expenses = df[df['Type'] == 'expense'].copy()
expenses['Date'] = pd.to_datetime(expenses['Date'])

monthly_spending = (
  expenses.groupby(expenses['Date'].dt.to_period('M'))['Amount']
  .sum()
  .sort_index()
)

print(montly_spending)
```
